# **INTRODUCCIÓN**

La ecuación de Hamilton-Jacobi es una herramienta fundamental en física y matemáticas, especialmente en la mecánica clásica y la teoría del control. Es considerada como la formulación más abstracta de la mecánica clásica. Sirve como puente entre esta y la mecánica cuántica. Su equivalente en la cuántica es la ecuación de Schrodinger.

Un aspecto importante de esta ecuación es que permite resolver de manera analítica algunos problemas en mecánica clásica. A través de esta formulación, se pueden encontrar soluciones exactas a problemas que de otro modo serían muy complicados de abordar. La formulación Hamiltoniana ofrece una formulación más general de la mecanica que las ecuaciones de movimiento de Newton o Lagrange.

Esta ecuación tiene aplicaciones en óptica,teoría del control, mecánica celeste, teoría de campos, entre otross.

En este trabajo analizaremos el comportamiento de una partícula bajo la influencia de un potencial no lineal usando la ecuación de Hamilton-Jacobi.

# **PLANTEAMIENTO DEL PROBLEMA**

Se tiene una partícula que se mueve en un campo gravitacional con un potencial que depende no linealmente de la posición. Suponiendo un potencial no central de la forma:

$$
\large V(x,y) = -\frac{GMm}{\sqrt{x^2 + y^2}} + k(x^2 + y^2)^n
$$

donde $n,k$ son constantes.

Se resolverá por el método de la ecuación de Hamilton-Jacobi y, a su vez, se verá cómo varía este sistema en función de las condiciones iniciales.


##**HAMILTONIANO DEL SISTEMA**



Es claro que para una particula bidimensional su Lagrangiano en coordenadas cartesianas quedaria de la siguiente forma:


$$\large L= \frac{m}{2}(\dot{x}^{2}+\dot{y}^{2})+\frac{GMm}{\sqrt{x^{2}+y^{2}}} - k(x^{2}+y^{2})^{n}$$

Veamos que por definición $ \large P_{x} \equiv \frac{∂}{∂\dot{x}}(L)= m\dot{x} $ y $ \large P_{y} \equiv \frac{∂}{∂\dot{y}}(L)= m\dot{y} $, lo que nos lleva a:


$$ \large \dot{x}=\frac{P_{x}}{m}, \dot{y}=\frac{P_{y}}{m}$$

Sabemos que el hamiltoniano se define de la siguiente forma:

$$ \large H(q_i, p_i, t) = \sum_{i} p_i \dot{q}_i - L(q_i, \dot{q}_i, t)$$

En nuestro caso, las coordenadas generalizadas son las coordenadas cartesianas $\large q_{i}=x,y$. Luego:

$$ \large \sum_{i} p_i \dot{q}_i=P_{x}\dot{x} +P_{y}\dot{y}= \frac{P_{x}^{2}}{m}+\frac{P_{y}^{2}}{m}$$

Por lo tanto:

$$ H = \frac{P_{x}^{2}}{m}+\frac{P_{y}^{2}}{m}-\frac{m}{2}(\dot{x}^{2}-\dot{y}^{2})-\frac{GMm}{\sqrt{x^{2}+y^{2}}} + k(x^{2}+y^{2})^{n}$$

$$H = \frac{P_{x}^{2}}{m}+\frac{P_{y}^{2}}{m}-\frac{m}{2}((\frac{P_{x}}{m})^{2}+(\frac{P_{y}}{m})^{2})-\frac{GMm}{\sqrt{x^{2}+y^{2}}} + k(x^{2}+y^{2})^{n} $$

Finalmente

$$ \large H=\frac{P_{x}^{2}}{2m}+\frac{P_{y}^{2}}{2m}-\frac{GMm}{\sqrt{x^{2}+y^{2}}} + k(x^{2}+y^{2})^{n}  $$


Es claro que el lagrangiano no tiene una dependencia explicita del tiempo entonces $H$ se conserva y además como el potencial no depende de las velocidades, se cumple que $H=E$ donde $E$ es la energia del sistema.

## **ECUACIÓN DE HAMILTON-JACOBI**

Ahora de la ecuación de Hamilton-Jacobi se tendria lo siguiente:


$$ \large
H\left(q_i, \frac{\partial S}{\partial q_i}, t\right) + \frac{\partial S}{\partial t} = 0$$


O que es igual:
$$ \large \frac{\partial S}{\partial t} = -E $$

Así:

$$ \large E= -\frac{\partial S}{\partial t}=
\frac{1}{2m}\left[ \left( \frac{\partial S}{\partial x} \right)^2 + \left( \frac{\partial S}{\partial y} \right)^2 \right] - \frac{G M m}{\sqrt{x^2 + y^2}} + k(x^2 + y^2)^n
$$

Suponiendo una solución de la forma:

$$ \large S=S^{'}(x,y)-Et$$

Es claro que:

$$ \large \frac{\partial S}{\partial x}= \frac{\partial S^{'}}{\partial x}$$  $$\large \frac{\partial S}{\partial y}= \frac{\partial S^{'}}{\partial y}$$ $$\large \frac{\partial S}{\partial t} = -E$$


Ahora es conveniente por tanto cambiar de coordenadas cartesianas a polares por el termino $$ \large x^{2}+y^{2}$$

Así:

$$ \large (x,y)---> (r,θ)$$

$$ \large S(x,y)----> S(r,θ) $$
$$ \large S'(x,y)----> S'(r,θ) $$




con las relaciones:

$$\large r^{2}= x^{2}+y^{2}$$  $$tan\frac{y}{x}=θ $$

y que es lo mismo que:

$$\large x=rcosθ$$, $$\large y=rsenθ$$


Es claro que como se desea pasar de un sistema coordenado a otro hay que ver como se transforman sus diferenciales y/o derivadas parciales respectos a las variables nuevas con las relaciones entre los sistemas coordenados.

Veamos lo siguiente:

$$\large \frac{\partial S}{\partial x}=\frac{\partial S}{\partial r}\frac{\partial r}{\partial x}+\frac{\partial S}{\partial θ}\frac{\partial θ}{\partial x} $$

$$\large \frac{\partial S}{\partial y}=\frac{\partial S}{\partial r}\frac{\partial r}{\partial y}+\frac{\partial S}{\partial θ}\frac{\partial θ}{\partial y} $$

y a su vez:

$$\large  \frac{\partial r}{\partial x}=cosθ $$
$$\large  \frac{\partial r}{\partial y}=senθ $$

$$\large  \frac{\partial θ}{\partial x}=\frac{-senθ}{r} $$
$$\large  \frac{\partial θ}{\partial y}=\frac{cosθ}{r} $$  



Así sustituyendo se tendria:

$$\large \frac{\partial S}{\partial x}=\frac{\partial S}{\partial r}cosθ-\frac{\partial S}{\partial θ}\frac{senθ}{r} $$

$$\large \frac{\partial S}{\partial y}=\frac{\partial S}{\partial r}senθ+\frac{\partial S}{\partial θ}\frac{cosθ}{r} $$


Para obtener la expresión cuadrática en la ecuación de Hamilton-Jacobi, calculamos:

$$\large
\left( \frac{\partial S}{\partial x} \right)^2 + \left( \frac{\partial S}{\partial y} \right)^2
$$

Sustituyendo las derivadas parciales:

$$\large
\left( \frac{\partial S}{\partial x} \right)^2 = \left( \frac{\partial S}{\partial r} \cos \theta - \frac{\partial S}{\partial \theta} \frac{\sin \theta}{r} \right)^2
$$

$$\large
\left( \frac{\partial S}{\partial y} \right)^2 = \left( \frac{\partial S}{\partial r} \sin \theta + \frac{\partial S}{\partial \theta} \frac{\cos \theta}{r} \right)^2
$$

Sumando estas expresiones:

$$\large
\left( \frac{\partial S}{\partial x} \right)^2 + \left( \frac{\partial S}{\partial y} \right)^2 = \left( \frac{\partial S}{\partial r} \right)^2 \left( \cos^2 \theta + \sin^2 \theta \right) + \frac{1}{r^2} \left( \frac{\partial S}{\partial \theta} \right)^2 \left( \sin^2 \theta + \cos^2 \theta \right)
$$

Utilizando $cos^{2}θ+sen^{2}θ=1 $:

$$\large
\left( \frac{\partial S}{\partial x} \right)^2 + \left( \frac{\partial S}{\partial y} \right)^2 = \left( \frac{\partial S}{\partial r} \right)^2 + \frac{1}{r^2} \left( \frac{\partial S}{\partial \theta} \right)^2
$$


Reemplazando en la ecuacion de hamilton Jacobi:

$$\large
E=\frac{1}{2m} \left[ \left( \frac{\partial S^{'}}{\partial r} \right)^2 + \frac{1}{r^2} \left( \frac{\partial S^{'}}{\partial \theta} \right)^2 \right] - \frac{G M m}{r} + k r^{2n}
$$



























Es adecuado notar lo sigueinte, el lagrangiano se transforma al cmabiarlo de sistema coordenado a esta forma:

$$\large L= \frac{m}{2}(\dot{r}^{2}+ r^{2}\dot{θ}^{2})+\frac{GMm}{r}-kr^{2n}$$

De acá:

$$\large P_{θ}≡\frac{∂}{∂\dot{θ}}L=cte$$

Así la coordenada $θ$ es ciclica lo que hace a $P_{θ}$ una constante de movimiento, este es el momento angular el cual se puede encontrar y/o fijar por condiciones inciales.


Denotando $$\large P_{θ}\equiv l$$



Suponiendo a su vez que se pueda realizar una separación de variables sobre $S^{'}$ de la forma:

$$\large S^{'}(r,θ
)=S_{r}(r)+ S_{\theta}(\theta) =S_{r}(r)+P_{\theta}\theta=S_{r}(r)+l\theta$$


Así:



$$\large
E=\frac{1}{2m} \left[ \left( \frac{\partial S_{r}}{\partial r} \right)^2 + \frac{1}{r^2} \left(  l\right)^2 \right] - \frac{G M m}{r} + k r^{2n}
$$

que es igual a:


$$\large (\frac{dS_{r}}{dr})^{2}+2m(-\frac{GMm}{r}+ kr^{2n})=2mE-(\frac{l}{r})^{2}                    $$

Ahora solo resta resolver esta ecuacion diferencial y reemplzar en $S$, notemos que se puede solucionar por integracion directa pues:

$$\large S_{r}=\int{\sqrt{2mE-(\frac{l}{r})^{2}- 2m(-\frac{GMm}{r}+ kr^{2n})}}dr $$


Así para algunos valores de n se podria solucionar dicha integral




Así por ejemplo para el caso de $n=1$

```python
r = sp.symbols('r',positive=True,real=True)
E, m, GM, k, l = sp.symbols('E m GM k l',positive=True,real=True)
n = sp.symbols('n',positive=True,real=True)
expr = 2*m*E + (r**2 * GM * m) / 2 - 2 * m * k * r**(2*n) - (l**2) / r**2
integrand = sp.sqrt(expr)
n_val = 1
integrand_with_n = integrand.subs(n, n_val)
integral = sp.integrate(integrand_with_n, r)
integral.doit()
```

La libreria de Simpy no resuelve dicha integral debido a que tiene muchas constantes no conocidas. Es adecuado, mencionar que dicha integral no se tiene que realizar de forma directa para resolver el problema, pues lo que realmente  interesa al usar la teoría de Hamilton-Jacobi son las derivadas parciales de $S$ respecto a $E,l$ (1)

Por tanto para dejarlo de forma general se tendria que la solución para $S$ seria:


$$\large S=\int{\sqrt{2mE-(\frac{l}{r})^{2}- 2m(-\frac{GMm}{r}+ kr^{2n})}}dr+l\theta -Et$$



Derivando esta expresión respecto a la constante $E$ y $l≡P_{θ}$.

$$\large \frac{∂S}{∂E}≡β_{1}=\int{\frac{mdr}{\sqrt{ 2mE-(\frac{l}{r})^{2}- 2m(-\frac{GMm}{r}+ kr^{2n}) }} } -t $$


$$\large \frac{∂S}{∂l}≡β_{2}=\int{\frac{-ldr }{r^{2}\sqrt{2mE-(\frac{l}{r})^{2}- 2m(-\frac{GMm}{r}+ kr^{2n}) }   } } + θ $$

donde $β_{1}, β_{2} $ son constantes que se encuentran a partir de las condiciones iniciales

   

Como ya se mencionó, la librería de simpy no es capaz de resolver dichas integrales de forma analítica. Así, estas integrales deben ser calculadas numéricamente para casos generales. Concentrémonos primero en el caso $n=0$. Este caso se convierte en el de una partícula en un potencial gravitacional kepleriano, pues, la constante que acompaña al termino $r^n$ no afecta la dinámica.




### **Caso $n=0$**

La integral toma la forma:

$$
\int \frac{mdr}{\sqrt{-2m \left(k - \frac{GMm}{r}\right) - \frac{l^2}{r^2} + 2Em}} = m \int \frac{dr}{\sqrt{-2m \left(k - \frac{GMm}{r}\right) - \frac{l^2}{r^2} + 2Em}}
$$

Resolviendo:
$$
\int \frac{dr}{\sqrt{-2m \left(k - \frac{GMm}{r}\right) - \frac{l^2}{r^2} + 2Em}}
$$

Sustituyendo $u = \frac{1}{r} \Rightarrow du = -\frac{dr}{r^2}$, así: $\frac{1}{r^2} = u^2$ o $\frac{1}{u^2} = r^2$:

$$
\int \frac{dr}{\sqrt{-2m \left(k - \frac{GMm}{r}\right) - \frac{l^2}{r^2} + 2Em}} = \int \frac{-du}{u^2 \sqrt{-l^2 u^2 + 2GMm^2 u - 2km + 2Em}}
$$

Resolviendo ahora:
$$
\int \frac{du}{u^2 \sqrt{-l^2 u^2 + 2GMm^2 u - 2km + 2Em}}
$$

Realizando la sustitución: $v = \frac{1}{u} \Rightarrow dv = -\frac{du}{u^2}$, además:
$$
u^2 = \frac{1}{v^2}
$$

Entonces:
$$
\int \frac{du}{u^2 \sqrt{-l^2 u^2 + 2GMm^2 u - 2km + 2Em}} = -\int \frac{vdv}{\sqrt{-2(k-E)mv^2 + 2GMm^2 v - l^2}}
$$

Reescribiendo $v$ como $\frac{GMm}{2(k-E)} - \frac{1}{4(k-E)m}(2GMm^2 - 4(k-E)mv)$, reemplazando:

$$
\int \left( \frac{GMm}{2(k-E)\sqrt{-2(k-E)mv^2 + 2GMm^2 v - l^2}} - \frac{2GMm^2 - 4(k-E)mv}{4(k-E)m \sqrt{-2(k-E)mv^2 + 2GMm^2 v - l^2}} \right) dr = \frac{1}{2(k-E)} \int \frac{(2k-2E)v - GMm}{\sqrt{-2(k-E)mv^2 + 2GMm^2 v - l^2}} dv + \frac{GMm}{2(k-E)} \int \frac{1}{\sqrt{-2(k-E)mv^2 + 2GMm^2 v - l^2}} dv
$$

Resolviendo:

$$
\int \frac{(2k-2E)v - GMm}{\sqrt{-2(k-E)mv^2 + 2GMm^2 v - l^2}} dv
$$

Con la sustitución $w = -2(k-E)mv^2 + 2GMm^2 v - l^2 \Rightarrow dw = (2GMm^2 - 4(k-E)mv)dv$:

$$
\int \frac{(2k-2E)v - GMm}{\sqrt{-2(k-E)mv^2 + 2GMm^2 v - l^2}} dv = \frac{-1}{2m} \int \frac{1}{\sqrt{w}} dw = -\frac{\sqrt{w}}{m} = \frac{-\sqrt{-2(k-E)mv^2 + 2GMm^2 v - l^2}}{m}
$$

Por otro lado:

$$
\int \frac{1}{\sqrt{-2(k-E)mv^2 + 2GMm^2 v - l^2}} dv = \int \frac{1}{\sqrt{\left(\sqrt{2(k-E)(-m)}v + \frac{GMm^2}{\sqrt{2(k-E)(-m)}}\right)^2 + \frac{G^2M^2m^2}{2(k-E)} - l^2}}
$$

Sustituyendo:
$$
w = \frac{GMm^2 - 2(k-E)mv}{\sqrt{2m(E-k) \left(\frac{G^2M^2m^2}{2(k-E)} - l^2\right)}}
$$

$$
\int \frac{1}{\sqrt{-2(k-E)mv^2 + 2GMm^2 v - l^2}} dv = \int \frac{1}{\sqrt{\left(\sqrt{2(k-E)(-m)}v + \frac{GMm^2}{\sqrt{2(k-E)(-m)}}\right)^2 + \frac{G^2M^2m^2}{2(k-E)} - l^2}} = \int \frac{-\sqrt{(E-k) \left(\frac{G^2M^2m^3}{2(k-E)} - l^2\right)}}{(\sqrt{2m}(k-E))(\sqrt{\left(-\frac{\sqrt{(E-k)(-m) \left(\frac{G^2M^2m^3}{k-E} - 2l^2\right)} w}{\sqrt{2m(k-E)}} + \frac{GMm^2}{2(k-E)(-m)} + \frac{GM(\sqrt{-m}m}{\sqrt{2(k-E)}}\right)^2 + \frac{G^2M^2m^3}{2(k-E)} - l^2} )}dw
=\frac{-\sqrt{2(E-k)}}{\sqrt{m(2k-2E)}}\int{\frac{1}{\sqrt{w²+1}}dw}=\frac{-\sqrt{2(E-k)}}{\sqrt{m(2k-2E)}}ln(\sqrt{w²+1}+w)=\frac{-\sqrt{2(E-k)}}{\sqrt{m(2k-2E)}}ln(\sqrt{\frac{(GMm²-2(k-E)mv)²}{2(E-k)m(\frac{G²M²m³}{2(k-E)}-l²)}+1}+\frac{(GMm²-2(k-E)mv)}{\sqrt{2m(E-k)(\frac{G²M²m³}{2(k-E)}-l²)}})$$

reeemplzando las integrales resueltas y deshaciendo la sustitución: $v=\frac{1}{u}$:


$$
\frac{1}{2(k-E)}\int{\frac{(2k-2E)v-GMm}{\sqrt{-2(k-E)mv^2 + 2GMm^2v - l^2}}} \, dv + \frac{GMm}{2(k-E)}\int{\frac{dv}{\sqrt{-2(k-E)mv^2 + 2GMm^2v - l^2}}}
= \frac{\sqrt{m(E-k)}GM \left( \ln \left( \sqrt{\frac{(GMm^2 - 2(k-E)mv)^2}{2(E-k)m \left( \frac{G^2M^2m^3}{2(k-E)} - l^2 \right)} + 1} \right) + \frac{GMm^2 - 2(k-E)mv}{\sqrt{2m(E-k) \left( \frac{G^2M^2m^3}{2(k-E)} - l^2 \right)}} \right)}{\sqrt{2}(k-E)(2k-2E)} - \frac{\sqrt{-2(k-E)mv^2 + 2GMm^2v - l^2}}{2(k-E)m}=\frac{\sqrt{\frac{2GMm²}{u} +\frac{2(k-E)m}{u²}+l²} }{2(k-E)m}-\frac{\sqrt(m(E-k)GM)}{\sqrt{2}(k-E)(2k-2E)}ln(\frac{GMm²-\frac{2(k-E)m}{u} }{\sqrt{2m(E-k(\frac{G²M²M³}{2(k-E)}-l²))}}+\sqrt{\frac{(GMm²-\frac{2(k-E)m}{u})²}{2(E-k)m(\frac{G²M²m³}{2(k-E)}-l²)}})
$$


Deshaciendo la sustitución $u=\frac{1}{r}$

$-\int{\frac{du}{u²\sqrt{-l²u²+2GMm²u -2km-2Em}}}=-\frac{\sqrt{m(E-k)}GM}{\sqrt{2}(k-E)(2k-2E)}ln(\sqrt{\frac{(GMm²-2(k-E)mr)²}{2(E-k)M(\frac{G²M²m³}{2(k-E)})+l²}+1}+\frac{GmM²-2(k-E)mr}{\sqrt{2m(E-k)(\frac{G²M²m³}{2(k-E)}-l²)}}) -frac{\sqrt{-2(k-E)mr²+2GMm²r-l²}}{2(k-E)m}$

Por lo tanto la solución quedaria de la forma:

$$ \int \frac{mdr}{\sqrt{-2m \left(k - \frac{GMm}{r}\right) - \frac{l^2}{r^2} + 2Em}}=-\frac{m^{\frac{3}{2}}GM\sqrt{E-k} }{2^{\frac{3}{2}}(k-E)² }(ln(|\sqrt{2}(E-k)\sqrt{\frac{G²M²m³-2(k-E)l²}{k-E}}\sqrt{2mr((k-E)r-GMm)+l²}+\sqrt{m(G²M²-2(k-E)l²)}(GMm-2(k-E)r )|) +\sqrt{2}(k-E)\sqrt{-2mr((k-E)r-GMm)-l²})+ cte$$






Por otro lado la otra integral también tiene forma analitica notemos que:

$$\int -\frac{l}{\sqrt{-2m \left(k - \frac{GMm}{r}\right) - \frac{l^{2}}{r^{2}} + 2Em} \, r^{2}} \, \mathrm{d}r= -l \int \frac{1}{\sqrt{2m \left(\frac{GMm}{r} - k\right) - \frac{l^{2}}{r^{2}} + 2Em} \, r^{2}} \, \mathrm{d}r$$

Resolviendo:

$$ \int \frac{1}{\sqrt{2m \left(\frac{GMm}{r} - k\right) - \frac{l^{2}}{r^{2}} + 2Em} \, r^{2}} \, \mathrm{d}r$$


Sustituyendo: $u = \frac{1}{r}--->$ $\mathrm{d}u = -\frac{1}{r^{2}} \, \mathrm{d}r $ , $\frac{1}{r^{2}} = u^{2}$

$$ \int \frac{1}{\sqrt{2m \left(\frac{GMm}{r} - k\right) - \frac{l^{2}}{r^{2}} + 2Em} \, r^{2}} \, \mathrm{d}r= -\int \frac{1}{\sqrt{-l^{2} u^{2} + 2GMm^{2} u - 2km + 2Em}} \, \mathrm{d}u$$

Resolviendo:

$$\int \frac{1}{\sqrt{-l^{2} u^{2} + 2GMm^{2} u - 2km + 2Em}} \, \mathrm{d}u= \int \frac{1}{\sqrt{-\left(lu - \frac{GMm^{2}}{l}\right)^{2} + \frac{G^{2} M^{2} m^{4}}{l^{2}} - 2km + 2Em}} \, \mathrm{d}u= \int \frac{1}{\sqrt{-\left(lu - \frac{GMm^{2}}{l}\right)^{2} + \frac{G^{2} M^{2} m^{4}}{l^{2}} - 2km + 2Em}} \, \mathrm{d}u$$

sustituyendo

$$v = \frac{\mathrm{i} \left(l^{2} u - GMm^{2}\right)}{l \sqrt{\frac{G^{2} M^{2} m^{4}}{l^{2}} - 2km + 2Em}}----> \mathrm{d}v = \frac{\mathrm{i}l}{\sqrt{\frac{G^{2} M^{2} m^{4}}{l^{2}} - 2km + 2Em}} \, \mathrm{d}u$$




$$ \int \frac{1}{\sqrt{-\left(lu - \frac{GMm^{2}}{l}\right)^{2} + \frac{G^{2} M^{2} m^{4}}{l^{2}} - 2km + 2Em}} \, \mathrm{d}u= \int -\frac{\mathrm{i} \sqrt{\frac{G^{2} M^{2} m^{4}}{l^{2}} - 2km + 2Em}}{l \sqrt{m \left(\frac{G^{2} M^{2} m^{3}}{l^{2}} - 2k + 2E\right) v^{2} + \frac{G^{2} M^{2} m^{4}}{l^{2}} - 2km + 2Em}} \, \mathrm{d}v= -\frac{\mathrm{i}}{l} \, \int \frac{1}{\sqrt{v^{2} + 1}} \, \mathrm{d}v$$

como

$$\int \frac{1}{\sqrt{v^{2} + 1}} \, \mathrm{d}v= \ln\left(\sqrt{v^{2} + 1} + v\right)$$


entonces:

$$-\frac{\mathrm{i}}{l} \, \int \frac{1}{\sqrt{v^{2} + 1}} \, \mathrm{d}v= -\frac{\mathrm{i} \ln\left(\sqrt{v^{2} + 1} + v\right)}{l}= -\frac{\mathrm{i} \ln\left(\sqrt{1 - \frac{\left(l^{2} u - GMm^{2}\right)^{2}}{l^{2} \left(\frac{G^{2} M^{2} m^{4}}{l^{2}} - 2km + 2Em\right)}} + \frac{\mathrm{i} \left(l^{2} u - GMm^{2}\right)}{l \sqrt{\frac{G^{2} M^{2} m^{4}}{l^{2}} - 2km + 2Em}}\right)}{l}$$

Así:


$$-\int \frac{1}{\sqrt{-l^{2} u^{2} + 2GMm^{2} u - 2km + 2Em}} \, \mathrm{d}u= \frac{\mathrm{i} \ln\left(\sqrt{1 - \frac{\left(l^{2} u - GMm^{2}\right)^{2}}{l^{2} \left(\frac{G^{2} M^{2} m^{4}}{l^{2}} - 2km + 2Em\right)}} + \frac{\mathrm{i} \left(l^{2} u - GMm^{2}\right)}{l \sqrt{\frac{G^{2} M^{2} m^{4}}{l^{2}} - 2km + 2Em}}\right)}{l}= \frac{\mathrm{i} \ln\left(\frac{\mathrm{i} \left(\frac{l^{2}}{r} - GMm^{2}\right)}{l \sqrt{\frac{G^{2} M^{2} m^{4}}{l^{2}} - 2km + 2Em}} + \sqrt{1 - \frac{\left(\frac{l^{2}}{r} - GMm^{2}\right)^{2}}{l^{2} \left(\frac{G^{2} M^{2} m^{4}}{l^{2}} - 2km + 2Em\right)}}\right)}{l}$$


Así:

$$\int -\frac{l}{\sqrt{-2m \left(k - \frac{GMm}{r}\right) - \frac{l^{2}}{r^{2}} + 2Em} \, r^{2}} \, \mathrm{d}r= -\mathrm{i} \ln\left(\left|\frac{\mathrm{i} \left(\frac{l^{2}}{r} - GMm^{2}\right)}{l \sqrt{\frac{G^{2} M^{2} m^{4}}{l^{2}} - 2km + 2Em}} + \sqrt{1 - \frac{\left(\frac{l^{2}}{r} - GMm^{2}\right)^{2}}{l^{2} \left(\frac{G^{2} M^{2} m^{4}}{l^{2}} - 2km + 2Em\right)}}\right|\right) + C$$


Por tanto al reemplazar los valores de las constantes en teoría se podrían conseguir dichas soluciones. Sin embargo, estas pueden conducir a errores numéricos, por tanto se propone resolverlo solo de forma computacional para algunos valores de $n$.

Al reemplazar las constantes $l,E,G,M,m $, se tiene que:


$β_{1}=f_{1}(r) - t$

$β_{2}=f_{2}(r) + θ$


De esto se puede despejar a $r$ y a  $θ$ en terminos del tiempo y las constantes



```python
def f1(E,G,M,m,k,l,r):
  num=m
  den=np.sqrt(2*m*E-((l/r)**2)-2*m*(-G*M/r+k))
  return num/den

def f2(E,G,M,m,k,l,r):
  num=-l
  den=r**2*np.sqrt(2*m*E-((l/r)**2)-2*m*(-G*M/r+k))
  return num/den
```

```python
#@title $f_1(r), f_2(r)$
import matplotlib.pyplot as plt
import numpy as np
rango=np.linspace(5.01,10,1000)

y=[f1(a1,a2,a3,a4,a5,a6,i) for i in rango]
y2=[f2(a1,a2,a3,a4,a5,a6,i) for i in rango]

fig,ax = plt.subplots(figsize=(7,4))
ax.plot(rango,y2,'--',label='$F_{2}$')
ax.plot(rango,y,'--',label='$F_{1}$')
ax.plot([5-0.01,5-0.01],[-4,10],'--r')
ax.set_xlabel('r [longitud]')
ax.grid()
ax.legend()
```

Notamos que ambas integrales tienen asintotas verticales en $r=5$, Lo que quiere decir que al reemplazar de forma arbitraria ciertos valores de las constantes, los valores de $r$ tendrán cierta restricción para que no aparezcan divisones por cero dentro de las integrales.

Teniendo en cuenta que se tienen que se tiene que imponer una condición en $r$(cota inferior, es decir $r>5$) entonces para integrar de forma numerica se tendria lo siguiente:


$$\Large β_{1}=\int_{r_{0}=5}^{r}{\frac{mdr}{\sqrt{ 2mE-(\frac{l}{r})^{2}- 2m(-\frac{GMm}{r}+ kr^{2n}) }} } -t $$


$$\Large β_{2}=\int_{r_{0}=5}^{r}{\frac{-ldr }{r^{2}\sqrt{2mE-(\frac{l}{r})^{2}- 2m(-\frac{GMm}{r}+ kr^{2n}) }   } } + θ $$

Como se mencionó con anterioridad las constates $β_{1},β_{2}$ se encuentran con las condiciones iniciales del sistema. Sin perdida de generalidad se puede suponer que en $t=0$ $\implies$ $r=10,θ=0$


$$β_{1}=\int_{r_{0}=5}^{10}{\frac{mdr}{\sqrt{ 2mE-(\frac{l}{r})^{2}- 2m(-\frac{GMm}{r}+ kr^{2n}) }} } $$


$$β_{2}=\int_{r_{0}=5}^{10}{\frac{-ldr }{r^{2}\sqrt{2mE-(\frac{l}{r})^{2}- 2m(-\frac{GMm}{r}+ kr^{2n}) }   } }$$

```python
from scipy.integrate import quad
import numpy as np
from scipy.interpolate import interp1d
from scipy.optimize import root_scalar
from scipy import interpolate
from scipy.optimize import brentq
from functools import partial

def int1(E,G,M,m,k,l,a,b):
  integrad=lambda r: f1(E,G,M,m,k,l,r)
  return quad(integrad,a,b)[0]
def int2(E,G,M,m,k,l,a,b):
  integrad=lambda r: f2(E,G,M,m,k,l,r)
  return quad(integrad,a,b)[0]
```

```python
#@title $E = 1, G = 1, M = 1000,m = 1,k = 1,l = 1$
def f1(E,G,M,m,k,l,r):
  num=m
  den=np.sqrt(2*m*E-((l/r)**2)-2*m*(-G*M/r+k*r))
  return num/den

def f2(E,G,M,m,k,l,r):
  num=-l
  den=r**2*np.sqrt(2*m*E-((l/r)**2)-2*m*(-G*M/r+k*r))
  return num/den

a1=1
a2=1
a3=1000
a4=1
a5=1
a6=1

import matplotlib.pyplot as plt
import numpy as np
rango=np.linspace(1,32,1000)
plt.title('Funciones a integrar para n=1/2')
y=[f1(a1,a2,a3,a4,a5,a6,i) for i in rango]
y2=[f2(a1,a2,a3,a4,a5,a6,i) for i in rango]
plt.plot(rango,y,'--',label='$F_{1}$')
plt.plot(rango,y2,'--',label='$F_{2}$')
plt.grid()
plt.legend()
plt.show()
```

# **INTRODUCCIÓN**

La relatividad general es una teoría propuesta por Albert Einstein en 1915. Es una de las mayores revoluciones de la física moderna. En esta teoría, la gravedad no es una fuerza en su sentido clásico, es una manifestación de la curvatura del espacio-tiempo que generan los cuerpos masivos como planetas y estrellas.

La métrica de Schwarzschild es una solución exacta de las ecuaciones de campo de Einstein en la relatividad general. Esta métrica describe el espacio tiempo alrededor de un cuerpo esférico y no cargado. Es una de las soluciones más simples y a la vez más fundamentales de la relatividad.

Un problema clásico que no podía ser explicado por la mecánica newtoniana era la precesión de la órbita de Mercurio. La precesión es el cambio en la orientación de la órbita de un planeta. La relatividad general proporcionó una explicación precisa para este fenómeno. La corrección relativista predise exactamente la cantidad de precesión adicional observada. Este resultado fue una de las primeras pruebas experimentales de la relatividad general.



En el siguiente trabajo se realizará una comparación entre las trayectorias de Mercurio teorizadas por la mecánica clásica y la relatividad general, además de de calcular la tasa de precesión de la órbita.

# **PLANTEAMIENTO DEL PROBLEMA**
De forma general, el diferencial de arco $ds$ se escribe como  $ds^{2}=g^{ik}dx_{i}dx^{k}$. Si usamos la métrica de Schwarzschild, el diferencial toma la forma:


$$\Large ds²=(1-\frac{r_{g}}{r})c²dt²-r²(sin²(θ)dϕ²+dθ²)-\frac{dr²}{1-\frac{r_{g}}{r}} $$

Donde $\large r_{g}=\frac{2kM}{c²}$, donde $M$ es la masa total de los cuerpos que producen el campo y $k$ es la constante de gravitación universal. $r_{g}$ se conoce comunmente como el radio de schwarzschild.

Recordemos que la ecuación de Hamilton-Jacobi viene dada por:

$$\large g^{ik}\frac{\partial S}{\partial x^{i}}\frac{\partial S}{\partial x^{k}}-m²c²=0$$

Para la métrica de Schawrzschild, esta ecuación toma la forma:

$$(1-\frac{r_{g}}{r})^{-1}(\frac{\partial S}{c∂t})^²-(1-\frac{r_{g}}{r})(\frac{\partial S}{∂r})^²-\frac{1}{r²}(\frac{∂S}{∂ϕ})^²-m²c²=0$$


Suponiendo una solución de la forma:

$$S=-Et+lϕ+S_{r}(r)\implies \frac{dS_{r}(r)}{dr}=\frac{∂S}{∂r}$$

despejando y reemplazando, se tendría:

$$ S_{r}=∫\sqrt{ (\frac{E^2}{c²}(1-\frac{r_{g}}{r}))^{-2} -(m²c²+\frac{l²}{r²})(1-\frac{r_{g}}{r})^{-1}} dr$$




Derivando respecto a las constantes $E$ y $M$ las ecuaciones para $r$ y $θ$ respectivamente, encontramos nuevas constantes correspondientes a las momentos de la transformación canónica $(r,\theta) -> (\beta_{1},\beta_{2})$.

$$ \large β_{1}\equiv\frac{∂S}{\partial E}= \frac{∂S_{r}}{\partial E}-t $$



Derivando se tiene que:

$$ \frac{∂S_{r}}{\partial E} = \int \frac{-2E/c^2(1-r_g/r)^{-2}}{2\sqrt{(\frac{E^2}{c²}(1-\frac{r_{g}}{r}))^{-2} -(m²c²+\frac{l²}{r²})(1-\frac{r_{g}}{r})^{-1}}} dr $$

Factorizando $E/c^2(1-r_g/r)^{-1}$$E/c^2(1-r_g/r)^{-1}$ en el numerador y $(1-r_g/r)^{-1}$$(1-r_g/r)^{-1}$ en el denominador, obtenemos:

$$ \frac{∂S_{r}}{\partial E} = -\frac{E}{c^2}\int \frac{dr}{(1-\frac{r_g}{r})\sqrt{(\frac{E}{c²})^2 - (1-\frac{r_g}{r})(m^2c^2 + \frac{M^2}{r^2})}} $$

Por tanto se tendria:


$$\large ct=\frac{E}{mc²}\int\frac{dr}{(1-\frac{r_{g}}{r})[(\frac{E}{mc²})²-(1+\frac{l²}{m²c²r²})(1-\frac{r_{g}}{r})]^{1/2}} $$


Analogamente:


$$\large ϕ=∫\frac{ldr}{r²\sqrt{\frac{E²}{c²}-(m²c²)+\frac{l²}{r²}(1-\frac{r_{g}}{r}) }} $$


Veamos que dicha integral tiene una forma:


$$ϕ=\int \frac{Mdr}{r²\sqrt{A-\frac{B}{r}-\frac{C}{r²}+\frac{D}{r³}}} $$

donde:

$A=\frac{E²}{c²}-m²c²$

$B=\frac{E²r_{g}}{c²}$

$C=l²$

$D=l²r_{g}$

sea $u=\frac{1}{r} \implies dr=-\frac{-du}{u²}$
Así la integral se transforma a:

$$\phi=-\int \frac{Mdr}{\sqrt{Au⁴-Bu³-Cu²+Du}}$$

Aproximando por series de potencias se tendria:

$$(\frac{du}{dϕ})^2 + u²=\frac{Gl}{l²}-ur_{g}+ \text{Lower order terms}$$


Notemos que de forma integral se podria realizar a su vez la siguiente aproximación:

$$\sqrt{\frac{E²}{c²}-m^2c²+\frac{M²}{r^2}(1-\frac{r_{g}}{r})}=\sqrt{A-\frac{M²r_{g}}{r³}}$$

Expandiendo la raíz cuadrada

$\sqrt{A-\frac{M²r_{g}}{r³}}≈\sqrt{A}(1-\frac{M²r_{g}}{2Ar³}) $

Así:

$\frac{1}{\sqrt{A-\frac{M²r_{g}}{r³}}}≈\frac{1}{\sqrt{A}}(1+\frac{M²r_{g}}{2Ar³}) $

Por tanto:

$$\large ϕ≈∫\frac{Mdr}{r²\sqrt{A}}(1+\frac{M²r_{g}}{2Ar³})=\frac{M}{\sqrt{A}}∫\frac{dr}{r²}+\frac{M³r_{g}}{2A^{\frac{3}{2}}}∫\frac{dr}{r⁵}$$


La primera integral sería solución a la órbita kepleriana.

$r(ϕ)=\frac{p}{1+ecosϕ}$

el otro termino sería la corrección que se le debe de realizar a dicha integral

Así la solución final seria:

$u(ϕ)=\frac{GM}{l²}(1+ecos(\phi-δϕ)) $
$\implies$

$\frac{1}{r(ϕ)}=\frac{GM}{l²}[1+ecos((1-δ)\phi)]$



La solución obtenida a partir de la mecánica clásica viene dada por:

$$\large r(θ)=\frac{a(1-e²)}{1+ecos(θ)}$$

donde $p=\frac{l²}{GM}$

Para terner una solución completa se debe de calcular el cambio en el perihelio partiendo de la integral de trayectoria se realiza un cambio de variable $r(r-r_{g})=r'² $ tal que $r-\frac{r_{g}}{2}≈r' $

Así, se obtiene la siguiente forma $M,E-->\frac{M²}{r'²},E'$ donde E' es la energia no relativista y al expandir en terminos de $\frac{r_{g}}{r'} $ se tendria:,


$$S_{r}=∫[(2E'm+\frac{E'²}{c²})+\frac{1}{r}(2m²Mk+4E'mr_{g})-\frac{1}{r²}(M²-\frac{3m²c²r_{g}²}{2}) ]^{1/2}dr $$

como:

$β_{2}\equivϕ+\frac{∂S_{r}}{∂l}$

$\implies $

$δϕ=-\frac{∂}{∂l}ΔS_{r}$

Así expandiendo en serie de potencias[1]

$$\large ΔS_{r}=ΔS_{r}^{(0)}-\frac{3m²c²r_{g}²}{4l}\frac{\partial ΔS_{r}^{0}}{\delta l}$$

donde $ΔS_{r}^{0}$ corresponde a el cambio en el movimiento en una elipse sin precesar.

es claro que:

$$\large -\frac{∂ ΔS_{r}^{0}}{∂l}=Δϕ^{0}=2π$$

por tanto:

$$\large Δ\phi=2ϕ+\frac{3πm²c²r_{g}²}{2l²}=2π+\frac{6ϕk²m²M²}{c²l²} $$

como se cumple que:

$$\large \frac{l²}{kMm²}=a(1-e²) $$

Se obtiene:

$$ \large δϕ=\frac{6π kM}{c²a(1-e²)}  $$

para mas información acerca de la deducción se recomienda[3]

```python
#@title **Mecánica clásica vs Mecánica relativista**

import numpy as np
import matplotlib.pyplot as plt

# Parámetros físicos para Mercurio
G = 6.67430e-11  # Constante de gravitación universal [m^3 kg^-1 s^-2]
M = 1.989e30     # Masa del Sol [kg]
c = 3e8          # Velocidad de la luz [m/s]
a = 5.791e10     # Semieje mayor de la órbita de Mercurio [m]
e = 0.2056       # Excentricidad de la órbita de Mercurio

# Radio de Schwarzschild
r_g = 2 * G * M / c**2  # Radio de Schwarzschild [m]

# Momento angular específico de Mercurio
L = np.sqrt(G * M * a * (1 - e**2))

# Ecuación de la órbita de Newton
def r_newton(theta):
    p = L**2 / (G * M)
    return p / (1 + e * np.cos(theta))

# Ecuación de la órbita relativista (Relatividad General)
def r_einstein(theta):
    p = L**2 / (G * M)
    delta = 3 * r_g * p / L**2
    return p / (1 + e * np.cos((1 - delta) * theta))

# Valores de theta
theta = np.linspace(0, 10 * np.pi, 10000)  # Simulamos más revoluciones para observar la precesión

# Calcular r para ambas teorías
r_n = r_newton(theta)
r_e = r_einstein(theta)

# Convertir a coordenadas cartesianas
x_n = r_n * np.cos(theta)
y_n = r_n * np.sin(theta)

x_e = r_e * np.cos(theta)
y_e = r_e * np.sin(theta)

# Graficar ambas órbitas
plt.figure(figsize=(7, 7))

ax = plt.axes()

# Setting the background color of the plot
# using set_facecolor() method
ax.set_facecolor("k")
plt.plot(x_n, y_n, label='Newton (Órbita Clásica)', color='yellow', linestyle='-', linewidth=6)
plt.plot(x_e, y_e,'--r', label='Einstein (Relatividad General)', linewidth=1)
plt.plot(0, 0, 'yo', label='Sol', markersize=10)
plt.title('Comparación de Órbitas de Mercurio: Newton vs Relatividad General')
plt.xlabel('x [m]')
plt.ylabel('y [m]')
plt.legend()
plt.grid(True)
plt.gca().set_aspect('equal', adjustable='box')
plt.xlim(-1e11, 1e11)
plt.ylim(-1e11, 1e11)

plt.show()
```

No hay una diferencia apreciable entre las órbitas.

Sin embargo al momento de pasar el tiempo, la variación del perihelio se va haciendo mas notable, haciendo así que la elipse tenga mas variación(en cuestión de rotación alrededor del sol).

```python
#@title **Precesión del perihelio**

import numpy as np
import matplotlib.pyplot as plt

# Parámetros físicos para Mercurio
G = 6.67430e-11  # Constante de gravitación universal [m^3 kg^-1 s^-2]
M = 1.989e30     # Masa del Sol [kg]
c = 3e8          # Velocidad de la luz [m/s]
a = 5.791e10     # Semieje mayor de la órbita de Mercurio [m]
e = 0.2056       # Excentricidad de la órbita de Mercurio

# Radio de Schwarzschild
r_g = 2 * G * M / c**2  # Radio de Schwarzschild [m]

# Momento angular específico de Mercurio
L = np.sqrt(G * M * a * (1 - e**2))

# Corrección relativista delta
p = L**2 / (G * M)  # Semi-latus rectum de la órbita
delta = 3 * r_g / p  # Corrección relativista que afecta la precesión

# Ecuación de la órbita relativista (Relatividad General)
def r_einstein(theta, precession=0):
    return p / (1 + e * np.cos(theta * (1 - delta) + precession))

# Valores de theta para una órbita completa
theta = np.linspace(0, 2 * np.pi, 1000)

# Calcular r para la primera órbita (t = 0)
r_e_1 = r_einstein(theta)

# Calcular r para una órbita después de mucho tiempo (t = 10^7 años, por ejemplo)
time_separation = 1e7  # Tiempo en años
precession = time_separation * delta  # Ajuste de la precesión en radianes
r_e_2 = r_einstein(theta, precession)

# Convertir a coordenadas cartesianas para ambas órbitas
x_e_1 = r_e_1 * np.cos(theta)
y_e_1 = r_e_1 * np.sin(theta)
x_e_2 = r_e_2 * np.cos(theta)
y_e_2 = r_e_2 * np.sin(theta)

# Graficar las dos órbitas
plt.figure(figsize=(7, 7))

ax = plt.axes()

# Setting the background color of the plot
# using set_facecolor() method
ax.set_facecolor("k")
plt.plot(x_e_1, y_e_1, label='Órbita inicial', color='red')
plt.plot(x_e_2, y_e_2, label='Órbita después de $10^7$ años', color='blue')
plt.plot(0, 0, 'yo', label='Sol', markersize=10)
plt.title('Precesión del Perihelio de Mercurio')
plt.xlabel('x [m]')
plt.ylabel('y [m]')
plt.legend()
plt.grid(True)
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

```python
#@title **Precesión del perihelio: Mecánica clásica vs mecánica relativista**
import numpy as np
import matplotlib.pyplot as plt

# Parámetros físicos para Mercurio
G = 6.67430e-11  # Constante de gravitación universal [m^3 kg^-1 s^-2]
M = 1.989e30     # Masa del Sol [kg]
c = 3e8          # Velocidad de la luz [m/s]
a = 5.791e10     # Semieje mayor de la órbita de Mercurio [m]
e = 0.2056       # Excentricidad de la órbita de Mercurio

# Radio de Schwarzschild
r_g = 2 * G * M / c**2  # Radio de Schwarzschild [m]

# Momento angular específico de Mercurio
L = np.sqrt(G * M * a * (1 - e**2))

# Corrección relativista delta
p = L**2 / (G * M)  # Semi-latus rectum de la órbita
delta = 3 * r_g / p  # Corrección relativista que afecta la precesión

# Ecuación de la órbita relativista (Relatividad General)
def r_einstein(theta, precession=0):
    return p / (1 + e * np.cos(theta * (1 - delta) + precession))

# Ecuación de la órbita newtoniana
def r_newton(theta):
    return p / (1 + e * np.cos(theta))

# Valores de theta para una órbita completa
theta = np.linspace(0, 2 * np.pi, 1000)

# Parámetros de la simulación
num_orbits = 5  # Número de órbitas a graficar
time_step = 1e6  # Paso temporal en años entre cada órbita

# Graficar múltiples órbitas separadas en el tiempo
plt.figure(figsize=(7, 7))
ax = plt.axes()
ax.set_facecolor("k")
for i in range(num_orbits):
    # Órbitas relativistas
    precession = i * time_step * delta
    r_e = r_einstein(theta, precession)
    x_e = r_e * np.cos(theta)
    y_e = r_e * np.sin(theta)
    plt.plot(x_e, y_e, label=f'Einstein después de {i*time_step:.1e} años')

    # Órbitas newtonianas (que no cambian con el tiempo)
    r_n = r_newton(theta)
    x_n = r_n * np.cos(theta)
    y_n = r_n * np.sin(theta)
    plt.plot(x_n, y_n, '--',color="white", label=f'Newton después de {i*time_step:.1e} años')

# Graficar el Sol
plt.plot(0, 0, 'yo', label='Sol', markersize=10)

# Configuración del gráfico
plt.title('Comparación de Órbitas: Newton vs Einstein (Precesión del Perihelio de Mercurio)')
plt.xlabel('x [m]')
plt.ylabel('y [m]')
plt.legend()
plt.grid(True)
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

Notemos que la variación en el angulo del perihelio se puede calcular

```python
#@title **Cálculo de la tasa de variación del perihelio**
import numpy as np

# Constantes
G = 6.67430e-11  # Constante de gravitación universal en m^3 kg^-1 s^-2
c = 299792458    # Velocidad de la luz en m/s
M_0 = 1.989e30   # Masa del Sol en kg
a = 5.791e10     # Semieje mayor de la órbita de Mercurio en metros
e = 0.2056       # Excentricidad de la órbita de Mercurio

# Cálculo de la precesión por revolución en radianes
delta_phi_per_revolution = (6 * np.pi * G * M_0) / (a * c**2 * (1 - e**2))

# Número de revoluciones por siglo
revolutions_per_century = 415.2

# Precesión total por siglo en radianes
delta_phi_per_century = delta_phi_per_revolution * revolutions_per_century

# Convertir a segundos de arco por siglo
precesion_arcsec_per_century = delta_phi_per_century * (180/np.pi) * 3600

# Imprimir el resultado
print(f"El valor del cambio en el ángulo del perihelio de Mercurio por siglo es de aproximadamente {precesion_arcsec_per_century:.2f} segundos de arco.")
```

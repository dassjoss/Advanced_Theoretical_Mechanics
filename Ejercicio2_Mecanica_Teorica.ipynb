# **INTRODUCCIÓN**

Un cuerpo cuyos tres momentos principales de inercia son distintos se llama peonza asimétrica [1]. Su estudio es de real importancia por sus diversas implicaciones y aplicaciones en campos relacionados a la física: El estudio de estos sistemas ayuda a los físicos a comprender fenómenos complejos como el caos y la estabilidad en movimientos rotacionales [2]. El comportamiento de cuerpos en rotación es fundamental en ingeniería mecánica y aeroespacial, pues, permite modelar el comportamiento de satélites, misiles y otros dispositivos que cambian de orientación en el espacio [3]. Además, los instrumentos de navegación se basan en principios derivados del estudio de objetos en rotación [4].

El problema de la peonza asimétrica resulta compleja debido a la no linealidad de las ecuaciones de movimiento, por lo que la descripción matemática del movimiento requiere de funciones especiales avanzadas como las funciones elípticas y las funciones theta de Jacobi. En este trabajo graficaremos las soluciones  analíticas del movimiento de un trompo asimétrico.

# **MARCO TEÓRICO**

## **Ecuaciones de euler**
Sea $\Large \frac{d \textbf{A}}{dt}$ la  tasa temporal de cambio de un vector $\large \textbf{A}$ respecto a un sistema fijo. Si el cambio del vector es meramente rotacional, podemos afirmar que:

$$\large \frac{d\textbf{A}}{dt} = \Omega  \times \textbf{A}$$

Donde $\large \Omega$ es la velocidad angular del vector $\textbf{A}$.

Cuando el cambio en $\large \textbf{A}$ se deba también a la traslación, entonces la ecuación toma la siguiente forma:

$$\large \frac{d\textbf{A}}{dt} = \frac{d\textbf{A}'}{dt}+\Omega  \times \textbf{A}$$

Donde $\Large \frac{d\textbf{A}'}{dt}$ es la tasa temporal de traslación del vector $\textbf{A}$.

Es bien sabido que $\Large \textbf{F} = \frac{d\textbf{P}}{dt}$ y $\Large \tau = \frac{d\textbf{M}}{dt}$, luego, podemos escribir estas ecuaciones de la siguiente forma:

\begin{equation}
\begin{matrix}
\Large \textbf{F} = \frac{d\textbf{P}'}{dt}+\Omega \times\textbf{P} & & & \Large \tau = \frac{d\textbf{M}'}{dt}+\Omega \times\textbf{M}\\
\end{matrix}
\end{equation}

Al escribir estas ecuaciones por componentes obtenemos:

\begin{equation}
\begin{matrix}
\large F_{1}=μ(\frac{dV_{1}}{dt} +Ω_{2}V_{3}-Ω_{3}V_{2})              & & & \large \tau_{1}=I_{1}\frac{dΩ_{1}}{dt} + (I_{3}-I_{2})Ω_{2}Ω_{3} \\
\large F_{2}=μ(\frac{dV_{2}}{dt} +Ω_{3}V_{1}-Ω_{1}V_{3})              & & & \large \tau_{2}=I_{2}\frac{dΩ_{2}}{dt} + (I_{1}-I_{3})Ω_{3}Ω_{1}\\
\large F_{3}=μ(\frac{dV_{3}}{dt} +Ω_{1}V_{2}-Ω_{2}V_{1})              & & & \large \tau_{3}=I_{3}\frac{dΩ_{3}}{dt} + (I_{2}-I_{1})Ω_{1}Ω_{2}
\end{matrix}
\end{equation}

Las ecuaciones de la segunda columna son las **ecuaciones de euler**




# **PLANTEAMIENTO DEL PROBLEMA**

La peonza asimétrica es un cuerpo rígido cuyos 3 momentos de inercia principales son distintos. Supongamos que sobre el sistema no actúan fuerzas externas ni torque. Las ecuaciones de euler quedan entonces de la siguiente forma:

$$\large \frac{dΩ_{1}}{dt}=  \frac{I_{2}-I_{3}}{I_{1}}Ω_{2}Ω_{3} $$

$$\large \frac{dΩ_{2}}{dt}=  \frac{I_{3}-I_{1}}{I_{2}}Ω_{1}Ω_{3} $$

$$\large \frac{dΩ_{3}}{dt}=  \frac{I_{1}-I_{2}}{I_{3}}Ω_{1}Ω_{2} $$

Supongamos que $I_{3}>I_{2}>I_{1}$

Suponiendo que estamos en los ejes principales del cuerpo tenemos:

$$\Large M_{i}=I_{i}Ω_{i} $$

Por lo que podemos escribir la energía y el momento angular como:

$$\large I_{1}^{2}Ω_{1}^{2}+I_{2}^{2}Ω_{2}^{2}+I_{3}^{2}Ω_{3}^{2}=M^{2}$$

$$\large I_{1}Ω_{1}^{2}+I_{2}Ω_{2}^{2}+I_{3}Ω_{3}^{2}=2E$$

Teniendo en cuenta que no hay fuerzas externas y que no hay torque, estas dos cantidades son conservadas. Estas cantidades pueden ser escritas en términos del vector $\textbf{M}$

$$\large \frac{M_1^2}{I_1}+\frac{M_2^2}{I_2}+\frac{M_3^2}{I_3} = 2E$$

$$\large M_1^2+M_2^2+M_3^2 = M^2$$

Despejando se tiene:

$$\large Ω_{1}^{2}=\frac{2EI_{3}-M^{2}}{I_{1}(I_{3}-I_{1})}  -Ω_{2}^{2}\frac{I_{2}(I_{3}-I_{2})}{I_{1}(I_{3}-I_{1})}               $$



$$  \large Ω_{3}^{2}=\frac{M^{2}-2EI_{1}}{I_{3}(I_{3}-I_{1})}  -Ω_{2}^{2}\frac{I_{2}(I_{2}-I_{1})}{I_{3}(I_{3}-I_{1})} $$

Reemplazando en las ecuaciones de euler tenemos:

$$  \large \frac{dΩ_{2}}{dt} = \frac{1}{I_2}\left(\frac{[(2EI_{3}-M^{2})-I_{2}(I_{3}-I_{2})Ω_{2}^{2}][(M^{2}-2EI_{1})-I_{2}(I_{2}-I_{1})Ω_{2}^{2}]}{I_{1}I_{3} }\right)^{1/2} $$


Suponiendo que $\large M^{2}>2EI_{2} $ y definiendo

$$\Large τ \equiv t\sqrt{\frac{(I_{3}-I_{2})(M^{2}-2EI_{1})}{I_{1}I_{2}I_{3}}  } $$

$$\Large s \equiv Ω_{2}\sqrt{\frac{I_{2}(I_{3}-I_{2})}{2EI_{3}-M^{2}}  } $$

Podemos definir un parametro $κ$ tal que $ 0<κ^{2}<1$ como

$$\large k^{2} \equiv \frac{(I_{2}-I_{1})(2EI_{3}-M^{2})}{(I_{3}-I_{2})(M^{2}-2EI_{1})} $$

Suponiendo una condición inicial $Ω_{2}(0)=0$,simplificando y despejando:

$$\large τ=\int_{0}^{s}\frac{ds'}{\sqrt{(1-s'^{2})(1-k^{2}s'^{2})}}$$


podemos invertir la integral anterior de la siguiente forma:


$$\Large s=f^{-1}(τ)≡sn(τ)$$

También conocida como una función eliptica de jacobi. Como $Ω_{1}(t) $ y $Ω_{3}(t) $ dependen de $Ω_{2}(t) $ resolviendo está integral e invirtiendo las anteriores relaciones se pueden encontrar las velocidades angulares.

Definiendo:

$$cn(τ)≡\sqrt{1-sn^{2}(τ)} $$

$$dn(τ)≡\sqrt{1-k^{2}sn^{2}(τ)} $$

Se tiene:

$$ \large Ω_{1}=\sqrt{\frac{2EI_{3}-M^{2}}{I_{1}(I_{3}-I_{1})}}  cn(τ) $$

$$ \large Ω_{2}=\sqrt{\frac{2EI_{3}-M^{2}}{I_{2}(I_{3}-I_{2})}}  sn(τ) $$

$$ \large Ω_{3}=\sqrt{\frac{M^{2}-2EI_{1}}{I_{3}(I_{3}-I_{1})}}  dn(τ) $$

Haciendo la sustitución $s'≡sin(α)$ se tiene:

$$\large ds'=cos(α)dα=\sqrt{1-s'^{2}}dα  $$

Por lo tanto:

$$\large τ=τ(Φ,k)=\int_{0}^{Φ}\frac{d\alpha}{\sqrt{1-k^{2}sin^{2}(α)}}  $$

donde $s=sen(Φ)$. \

Se tiene entonces que:

$$\large τ(Φ,k)=\int^{sen^{-1}(s)≡Φ}_{0} \frac{dα}{\sqrt{1-k^{2}sin^{2}(α)}}$$

considerando $k$ fijo:

$$\large f^{-1}(τ)=sen^{-1}(s)=Φ=f^{-1}\left(\int^{sen^{-1}(s)≡Φ}_{0} \frac{dα}{\sqrt{1-k^{2}sin^{2}(α)}} \right) $$

Así:

$$\large s=sen\left( f^{-1}\left(\int^{sen^{-1}(s)≡Φ}_{0} \frac{dα}{\sqrt{1-k^{2}sin^{2}(α)}} \right)\right) $$

Así se tiene que:

$$\large s=sn(τ)=sen\left( f^{-1}\left(\int^{sen^{-1}(s)}_{0} \frac{dα}{\sqrt{1-k^{2}sin^{2}(α)}} \right) \right)$$



Por lo tanto:


$$\large Ω_{2}(t)=\sqrt{\frac{2EI_{3}-M^{2}}{I_{2}(I_{3}-I_{2})} }sn(τ(t))=\sqrt{\frac{2EI_{3}-M^{2}}{I_{2}(I_{3}-I_{2})} }sn\left(t\sqrt{\frac{(I_{3}-I_{2})(M^{2}-2EI_{1})}{I_{1}I_{2}I_{3}}  }\right) $$
















Se imponen las siguientes condiciones para que no aparezcan términos imaginarios:

$$\large 2EI_{3}-M^{2}>0$$

$$\large M^{2}-2EI_{1} >0$$

$$I_{3}\neq I_{2} \neq I_{1} $$

Y además teniendo en cuenta las relaciones citadas

$$I_{3}>I_{2}>I_{1}$$
$$M^{2}>2EI_{2}$$

Podemos escoger las siguientes constantes para graficar las soluciones:

Condiciones iniciales: $\large M=1,E=\frac{1}{2}.$

Momentos de inercia del sistema: $\large I3=2,I2=9/10,I1=3/10 $










# **CÁLCULO ANALÍTICO**

```python
#@title **Librerias y funciones**
import scipy as sp
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import ellipj as f
from scipy.special import ellipkinc as func
from mpmath import *



mp.dps=25;mp.pretty=True

def inver(x,m):
  return f(x,m)[3] ## saca la amplitud


def Constantes(I1,I2,I3,M,E):
  a=np.sqrt((2*E*I3-M**2)/(I2*(I3-I2)))
  b=np.sqrt(      ((I3-I2)*(M**2-2*E*I1)) / (I1*I2*I3)     )
  return a,b

def k2(I1,I2,I3,M,E):
  return (((I2-I1)*(2*E*I3-M**2)) /((I3-I2)*(M**2-2*E*I1)))


def omega1(I1,I2,I3,M,E,t):
  a1=np.sqrt((2*E*I3-M**2)/(I1*(I3-I1)))
  b1=np.sqrt(((I3-I2)*(M**2-2*E*I1)) / (I1*I2*I3))
  m=k2(I1,I2,I3,M,E)
  arg=(t*b1,m)
  valor=f(*arg)
  return a1*valor[1]


def omega2(I1,I2,I3,M,E,t):
  a1,b1=Constantes(I1,I2,I3,M,E)
  m=k2(I1,I2,I3,M,E)
  arg=(t*b1,m)
  valor=f(*arg)
  return a1*valor[0]

def omega3(I1,I2,I3,M,E,t):
  a1=np.sqrt((M**2-2*E*I1)/(I3*(I3-I1)))
  b1=np.sqrt(((I3-I2)*(M**2-2*E*I1)) / (I1*I2*I3))
  m=k2(I1,I2,I3,M,E)
  arg=(t*b1,m)
  valor=f(*arg)
  return a1*valor[2]


def q(mi):
  return qfrom(m=mi)


def v1(z,q):
  return jtheta(1,z,q)

def v2(z,q):
  return jtheta(2,z,q)

def v3(z,q):
  return jtheta(3,z,q)

def v4(z,q):
  return jtheta(4,z,q)

def t(u,v):
  return u/(v**2)

## las que se definirán tienen agregado de una vez el escalar a1
def cn(I1,I2,I3,M,E,t):
  a1=np.sqrt((2*E*I3-M**2)/(I1*(I3-I1)))
  b1=np.sqrt(((I3-I2)*(M**2-2*E*I1)) / (I1*I2*I3))
  m=k2(I1,I2,I3,M,E)
  u=t*b1
  qs=q(m)
  ts=u/(v3(0,qs)**2)
  return a1*((v4(0,qs)*v2(ts,qs))/(v2(0,qs)*v4(ts,qs)))


def sn(I1,I2,I3,M,E,t):
  a1,b1=Constantes(I1,I2,I3,M,E)
  m=k2(I1,I2,I3,M,E)
  qs=q(m)
  u=t*b1
  ts=u/(v3(0,qs)**2)
  return a1*((v3(0,qs)*v1(ts,qs))/(v2(0,qs)*v4(ts,qs)))

def dn(I1,I2,I3,M,E,t):
  b1=Constantes(I1,I2,I3,M,E)[1]
  a1=np.sqrt((M**2-2*E*I1)/(I3*(I3-I1)))
  m=k2(I1,I2,I3,M,E)
  qs=q(m)
  u=t*b1
  ts=u/(v3(0,qs)**2)
  return a1*((v4(0,qs)*v3(ts,qs))/(v3(0,qs)*v4(ts,qs)))





```

```python
#@title $\Omega_1(t), \Omega_2(t),\Omega_3(t)$ **en términos de las funciones elípticas**

dominio=np.linspace(1,10,100) #10 segundos
rango1=[omega1(3/10,9/10,2,1,1/2,t) for t in dominio]
rango2=[omega2(3/10,9/10,2,1,1/2,t) for t in dominio]
rango3=[omega3(3/10,9/10,2,1,1/2,t) for t in dominio]
rangon = [rango1,rango2,rango3]

fig,ax = plt.subplots(1,3,figsize=(20,5))
colores=["red","purple","green"]
for i in range(3):
  ax[i].plot(dominio, rangon[i],label='$\Omega{0}(t)$'.format(str(i+1)),color=colores[i])
  ax[i].set_xlabel("Tiempo [s]")
  ax[i].set_ylabel("$\Omega$"+str(i+1))
  ax[i].grid()
  ax[i].legend()
```

## **LAS FUNCIONES THETA DE JACOBI**

Cuando se desea obtener resultados numéricos de la funciones elipticas de Jacobi, resultado pues, más sencillo resolverlas con ayuda de ciertas funciones auxiliares, las cuales son conocidas como las funciones theta.
Dichas funciones se encuentran definidas de la siguiente forma $[5]$:

$$ϑ(z,q)=∑_{n=-∞}^{∞}(-1)^{n}q^{n^{2}}e^{2niz}$$

es evidente que:

$$\vartheta(z, q) = 1 + 2 \sum_{n=1}^{\infty} (-1)^n q^n \cos(2nz)$$

y que $ϑ(z+π,q)=ϑ(z,q)$ \

por lo tanto:

$$\vartheta\left(z + \pi\tau, q\right) = \sum_{n=-\infty}^{\infty} (-1)^n q^n q^{2n} e^{2niz}=-q^{-1} e^{-2iz} \sum_{n=-\infty}^{\infty} (-1)^{n+1} q^{(n+1)^{2}} e^{2(n+1)iz}$$

$$\vartheta(z + \pi\tau, q) = -q^{-1} e^{-2iz} \vartheta(z, q)$$

Así, se dice que es una función quasi doble periodica sobre z.El efecto de incrementar z por $π$ o$πτ$ tiene el mismo efecto que multiplizar $\vartheta(z,q)$ por 1 o por $-q^{-1} e^{-2iz} $ los cuales son llamados los factores multiplicativos de la periodicidad asociada a los periodos  $π$ y $πτ$ respectivamente.

Se acostumbrá escribir

$ϑ_{4}(z,q) $ en lugar de $ϑ(z,q)$

Se define las funciones $ϑ_{3}(z,q)$,$ϑ_{2}(z,q)$,$ϑ_{1}(z,q)$ por la siguientes ecuaciónes:

$$\vartheta_3(z, q) = \vartheta_4\left(z + \frac{1}{2}\pi, q\right) = 1 + 2 \sum_{n=1}^{\infty} q^{n^{2}} \cos(2nz)$$

$$\vartheta_1(z, q) = -i e^{{iz} + \frac{1}{4} \pi i \tau} \vartheta_4\left(z + \frac{1}{2}\pi \tau, q\right)= -i\sum_{n=0}^{\infty} -1^{n}q^{(n+ \frac{1}{2})^{2}} e^{(2n+1)iz}$$

y así:

$$\vartheta_1(z, q)= 2\sum_{n=0}^{\infty} -1^{n}q^{(n+ \frac{1}{2})^{2}} sin((2n+1)z) $$

y por último definiendo $ϑ_2(z,q)$ por la siguiente ecuacion:

$$\vartheta_2(z, q) = \vartheta_1\left(z + \frac{1}{2}\pi, q\right) = 2\sum_{n=0}^{\infty} q^{(n+ \frac{1}{2})^{2}} cos((2n+1)z)$$

Se pueden reescribir las funciones $sn(τ)$,$cn(τ)$,$dn(τ)$ de la siguiente forma $[5]$:


\begin{align*}
\mathrm{sn}(τ,m=k^{2}) &= \frac{\vartheta_3(0,q)}{\vartheta_2(0,q)}
    \frac{\vartheta_1(t,q)}{\vartheta_4(t,q)} \\
\mathrm{cn}(τ,m=k^{2}) &= \frac{\vartheta_4(0,q)}{\vartheta_2(0,q)}
    \frac{\vartheta_2(t,q)}{\vartheta_4(t,q)} \\
\mathrm{dn}(τ,m=k^{2}) &= \frac{\vartheta_4(0,q)}{\vartheta_3(0,q)}
    \frac{\vartheta_3(t,q)}{\vartheta_4(t,q)},
\end{align*}



donde $q=q(m)$ y además:

\begin{align*}
\vartheta_1(z,q) &= 2 q^{1/4} \sum_{n=0}^{\infty}
  (-1)^n q^{n^2+n} \sin((2n+1)z) \\
\vartheta_2(z,q) &= 2 q^{1/4} \sum_{n=0}^{\infty}
  q^{n^2 + n} \cos((2n+1)z) \\
\vartheta_3(z,q) &= 1 + 2 \sum_{n=1}^{\infty}
  q^{n^2} \cos(2 n z) \\
\vartheta_4(z,q) &= 1 + 2 \sum_{n=1}^{\infty}
  (-q)^{n^2} \cos(2 n z)
\end{align*}

# **PERIODO**


El periodo se puede ver en terminos de la integral elíptica de Jacobi de primera especie. Sea $K$ definida como sigue:


$$K=\int_{0}^{1}\frac{ds'}{\sqrt{(1-s'^{2})(1-k^{2}s'^{2})}}=\int^{\frac{1π}{2}}_{0} \frac{dα}{\sqrt{1-k^{2}sin^{2}(α)}}$$

Según [1] se tiene que el periodo es:

$$\large T=4K\sqrt{\frac{I_{1}I_{2}I_{3}}{(I_{3}-I_{2})(M^{2}-2EI_{1})}}$$





```python
#@title **Cálculo del periodo $T$**
from scipy.special import ellipk as Ks

def periodo(I1,I2,I3,M,E):
  m=k2(I1,I2,I3,M,E)
  constante=np.sqrt(  (I1*I2*I3)/((I3-I2)*((M**2)-2*E*I1))   )
  return 4*Ks(m)*constante

print("El periodo teniendo en cuenta los parametros I1=3/10,I2=9/10,I3=2,M=1,E=1/2 seria",np.round((periodo(3/10,9/10,2,1,1/2)),3),'s')
```

Otra forma de verificar la validez de las funciones graficadas es verificar el comportamiento del vector $\textbf{M}$.

Recordemos que:

$$\large \frac{M_1^2}{I_1}+\frac{M_2^2}{I_2}+\frac{M_3^2}{I_3} = 2E$$

$$\large M_1^2+M_2^2+M_3^2 = M^2$$

Si elegimos como ejes $M_1,M_2,M_3$, las relaciones anteriores son las ecuaciones de una esfera de radio $M$ y un elipsoide de semiejes $\sqrt{2EI_1}, \sqrt{2EI_2},\sqrt{2EI_3}$. Cuando el vector M se mueve, su extremo se desplaza a lo alrgo de la línea de intersección de estas superficies [1].

La existencia de una intersección está asegurada por la siguiente desigualdad ya mencionada:

$$\Large 2EI_1 <M^2<2EI_3$$

Veamos algunas posibles trayectorias del vector $\textbf{M}$ en la superficie del elipsoide para distintos valores de $M$.


```python
#@title **Trayectorias del vector $\textbf{M}$**
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from scipy.optimize import fsolve
rx = np.sqrt(I1)
ry = np.sqrt(I2)
rz = np.sqrt(I3)
a, b, c = rx, ry, rz  # Semiejes del elipsoide

u = np.linspace(0, 2 * np.pi, 100)
v = np.linspace(0, np.pi, 100)
u, v = np.meshgrid(u, v)

xe = a * np.cos(u) * np.sin(v)
ye = b * np.sin(u) * np.sin(v)
ze = c * np.cos(v)

xc = 1 * np.cos(u) * np.sin(v)
yc = 1 * np.sin(u) * np.sin(v)
zc = 1 * np.cos(v)


fig = plt.figure(figsize=(14,8))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(xe, ye, ze, color='darkkhaki', alpha=0.26)

def intersection(R, a, b, c, theta):
    def equations(p):
        u, v = p
        eq1 = (a * np.cos(u) * np.sin(v))**2 + (b * np.sin(u) * np.sin(v))**2 + (c * np.cos(v))**2 - R**2
        eq2 = u - theta
        return (eq1, eq2)

    v0 = np.pi / 2
    return fsolve(equations, (theta, v0),xtol=1e-13)

# Radios de las esferas
radii = [1.8*rx,1.5*rx,1*rx,2*rx,2.5*rx]

for R in radii:
    theta_values = np.linspace(0, 2 * np.pi, 1000)
    curve = np.array([intersection(R, a, b, c, theta) for theta in theta_values])

    u_vals = curve[:, 0]
    v_vals = curve[:, 1]

    x_inter = a * np.cos(u_vals) * np.sin(v_vals)
    y_inter = b * np.sin(u_vals) * np.sin(v_vals)
    z_inter = c * np.cos(v_vals)

    ax.plot(x_inter, y_inter, z_inter,'.', label=f'M={np.round(R,3)}')

for R in radii:

  num_points = 100
  u = np.linspace(0, 2 * np.pi, num_points)
  v = np.linspace(0, np.pi, num_points)
  X = R * np.outer(np.cos(u), np.sin(v))
  Y = R * np.outer(np.sin(u), np.sin(v))
  Z = R* np.outer(np.ones_like(u), np.cos(v))

#ax.plot_surface(xc, yc, zc, color='red', alpha=0.3)


ax.set_xlabel("M1")
ax.set_ylabel("M2")
ax.set_zlabel("M3")

ax.set_box_aspect([ub - lb for lb, ub in (getattr(ax, f'get_{a}lim')() for a in 'xyz')])


ax.legend()
plt.show()

print('')
```

Si nuestras funciones están graficadas correctamente, entonces el vector $\textbf{M}(t)$ debe seguir la trayectoria definida por la intersección de una esfera de radio $M = 1$ y un elipsoide de semiejes $\sqrt{2EI_1}, \sqrt{2EI_2},\sqrt{2EI_3}$. Veamos como se comporta este vector en el tiempo:
